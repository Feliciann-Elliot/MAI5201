```python
from huggingface_hub import upload_file
from transformers import BertConfig

config = BertConfig(
    vocab_size=30522,
    hidden_size=768,
    num_hidden_layers=12,
    num_attention_heads=12,
    intermediate_size=3072,
    num_labels=2,
    id2label={0: "NEGATIVE", 1: "POSITIVE"},
    label2id={"NEGATIVE": 0, "POSITIVE": 1},
)

config.save_pretrained("./dummy_model_config")

upload_file(
    path_or_fileobj="./dummy_model_config/config.json",
    path_in_repo="config.json",
    repo_id="darantnel/dummy-model-v2",
    commit_message="Upload config.json"
)
```

```python
# First handle any unstaged changes
try:
    # Initialize git-lfs
    subprocess.run(["git", "lfs", "install"], cwd="./dummy_model_folder", check=True)
    
    # Track the specific file types with git-lfs
    lfs_track_patterns = ["*.bin", "*.json", "*.model", "*.safetensors"]
    for pattern in lfs_track_patterns:
        subprocess.run(["git", "lfs", "track", pattern], cwd="./dummy_model_folder", check=True)
    
    # Stage all files including .gitattributes
    repo.git_add(".gitattributes", auto_lfs_track=True)
    repo.git_add("*.json", auto_lfs_track=True)
    repo.git_add("*.model", auto_lfs_track=True)
    repo.git_add("*.safetensors", auto_lfs_track=True)
    
    # Commit the changes
    repo.git_commit("Add model files and LFS tracking")
    
    # Now safe to pull
    repo.git_pull()
    
    # Push changes
    repo.git_push()
    
    # Add a tag if needed
    repo.git_tag(tag_name="v1.0.0", message="Initial model version")
    repo.git_push(tags=True)
    
    print("Successfully pushed all changes")
    
except Exception as e:
    print(f"Error occurred: {str(e)}")
    # Print git status for debugging
    status = subprocess.run(
        ["git", "status"],
        cwd="./dummy_model_folder",
        capture_output=True,
        text=True
    )
    print("\nGit status:")
    print(status.stdout)
```
